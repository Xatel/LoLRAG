```python
from bs4 import BeautifulSoup
import requests
    
URL = "https://wiki.leagueoflegends.com/en-us/List_of_champions"
BASE_URL = "https://wiki.leagueoflegends.com"
```

```python
response = requests.get(URL)
html = response.text
soup = BeautifulSoup(html, "html.parser")
```

```python
list_of_champions = soup.find("table", {"class": "article-table"}).find("tbody").find_all("tr")
list_of_champion_hrefs = []

# Skip the first row since it is the header
for i in list_of_champions[1:]:
    href = i.find_all("td")[0].find("a").get("href")
    list_of_champion_hrefs.append(href)
```

```python
class ChampionScraper:
    def __init__(self, champion_href):
        self.champion_url = BASE_URL + champion_href
        self.champion_name = champion_href.split("/")[-1]
        self.champion_html = None
        self.champion_soup = None
    
    def scrape_champion_page(self):
        response = requests.get(self.champion_url)
        self.champion_html = response.text
        self.champion_soup = BeautifulSoup(self.champion_html, "html.parser")
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

options = Options()
# options.add_argument("--headless")

driver = webdriver.Chrome(options=options)
driver.get(URL)

wait = WebDriverWait(driver, 10)

table = wait.until(EC.presence_of_element_located((By.CLASS_NAME, "article-table")))
list_of_champions = table.find_elements(By.CSS_SELECTOR, "tbody tr")
```

```python
champion_names = []
for i in list_of_champions:
    champion_names.append(i.find_elements(By.TAG_NAME, "td")[0].get_attribute("data-sort-value"))
```
